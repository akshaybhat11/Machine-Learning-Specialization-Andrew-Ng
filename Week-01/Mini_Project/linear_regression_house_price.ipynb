# House Price Prediction using Linear Regression

**Week 1 Mini Project — Machine Learning Specialization (Andrew Ng)**

This notebook implements linear regression **from scratch** (no scikit-learn) using:
- Model function: `f(x) = wx + b`
- Cost function (Mean Squared Error)
- Gradient Descent to learn `w` and `b`

**Goal:** Predict house price from house size.

## Step 1: Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Create results folder if it doesn't exist (for saving plots)
os.makedirs('results', exist_ok=True)
```

## Step 2: Create the Dataset

A small dataset: house size (in 1000 sq ft) vs. price (in $1000s).

```python
# House size (in 1000 sq ft)
x_train = np.array([1.0, 1.2, 1.5, 1.8, 2.0, 2.3, 2.5, 2.8, 3.0, 3.2])

# House price (in $1000s)
y_train = np.array([150, 168, 190, 220, 240, 260, 285, 310, 330, 350])

print(f'Number of training examples: {x_train.shape[0]}')
print('x_train:', x_train)
print('y_train:', y_train)
```

## Step 3: Visualize the Data

Plot house size vs. price to see the relationship before modeling.

```python
plt.figure(figsize=(6,4))
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual data')
plt.title('House Size vs. Price')
plt.xlabel('House Size (1000 sq ft)')
plt.ylabel('Price ($1000s)')
plt.legend()
plt.grid(True)
plt.show()
```

## Step 4: Define the Linear Regression Model

$$f_{w,b}(x) = wx + b$$

```python
def compute_model_output(x, w, b):
    """
    Computes the prediction of a linear model.
    Args:
        x (ndarray): input features
        w, b (scalar): model parameters
    Returns:
        f_wb (ndarray): predicted values
    """
    return w * x + b
```

## Step 5: Define the Cost Function

The cost function measures how well the model fits the data (Mean Squared Error):

$$J(w,b) = \frac{1}{2m}\sum_{i=0}^{m-1}(f_{w,b}(x^{(i)}) - y^{(i)})^2$$

```python
def compute_cost(x, y, w, b):
    """
    Computes the cost function for linear regression.
    """
    m = x.shape[0]
    predictions = compute_model_output(x, w, b)
    cost = np.sum((predictions - y) ** 2) / (2 * m)
    return cost
```

## Step 6: Implement Gradient Descent

Gradient descent updates `w` and `b` to minimize the cost function:

$$w = w - \alpha \frac{\partial J}{\partial w}, \quad b = b - \alpha \frac{\partial J}{\partial b}$$

```python
def compute_gradient(x, y, w, b):
    """
    Computes the gradient of the cost function w.r.t. w and b.
    """
    m = x.shape[0]
    predictions = compute_model_output(x, w, b)
    error = predictions - y

    dj_dw = np.sum(error * x) / m
    dj_db = np.sum(error) / m

    return dj_dw, dj_db


def gradient_descent(x, y, w_init, b_init, alpha, num_iters):
    """
    Performs gradient descent to learn w and b.
    Returns final w, b, and history of cost values.
    """
    w = w_init
    b = b_init
    cost_history = []

    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(x, y, w, b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        cost = compute_cost(x, y, w, b)
        cost_history.append(cost)

        if i % 100 == 0:
            print(f'Iteration {i:4}: Cost {cost:8.2f}, w: {w:.4f}, b: {b:.4f}')

    return w, b, cost_history
```

## Step 7: Train the Model

```python
# Initialize parameters
w_init = 0
b_init = 0

# Hyperparameters
alpha = 0.05      # learning rate
num_iters = 1000  # number of iterations

# Run gradient descent
w_final, b_final, cost_history = gradient_descent(
    x_train, y_train, w_init, b_init, alpha, num_iters
)

print(f'\nFinal parameters: w = {w_final:.4f}, b = {b_final:.4f}')
```

## Step 8: Plot the Regression Line

Visualize how well the learned line fits the training data.

```python
plt.figure(figsize=(6,4))
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual data')

predicted = compute_model_output(x_train, w_final, b_final)
plt.plot(x_train, predicted, c='b', label='Linear regression fit')

plt.title('House Price Prediction - Linear Regression Fit')
plt.xlabel('House Size (1000 sq ft)')
plt.ylabel('Price ($1000s)')
plt.legend()
plt.grid(True)

plt.savefig('results/regression_line.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Step 9: Plot Cost vs. Iterations

This shows gradient descent converging — the cost should decrease over iterations.

```python
plt.figure(figsize=(6,4))
plt.plot(range(num_iters), cost_history, c='g')
plt.title('Cost vs. Iterations')
plt.xlabel('Iteration')
plt.ylabel('Cost J(w,b)')
plt.grid(True)

plt.savefig('results/cost_vs_iterations.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Step 10: Predict the Price of a New House

Use the trained model to predict the price of a house of a given size.

```python
new_size = 2.6  # in 1000 sq ft
predicted_price = compute_model_output(new_size, w_final, b_final)

print(f'Predicted price for a {new_size*1000:.0f} sq ft house: ${predicted_price*1000:.2f}')
```
